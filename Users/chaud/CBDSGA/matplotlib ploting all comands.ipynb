```python
import pandas as pd

df_sales = pd.read_excel("linechart.xlsx")
df_sales.head()
```

```python
from matplotlib import pyplot as plt
plt.figure(figsize=(12,5))
plt.plot(df_sales["Quarter"], df_sales["Fridge"], color="blue",label="Fridge")
plt.plot(df_sales["Quarter"], df_sales["Dishwasher"], color="green", label="Dishwasher")
plt.plot(df_sales["Quarter"], df_sales["Working machine"], color="orange", label="Washing machine")

plt.title("product Sales")
```

```python
total_sales = df_sales[["Fridge","Dishwasher","Working machine"]].sum()
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales, labels=total_sales.index, autopct="%1.3f%%", explode=(0.1,0,0), shadow=True, startangle=140)


plt.show()
```

```python
df_sales.plot(kind="bar",x="Quarter")
plt.xticks(rotation=45)

plt.show()
```
