```python
numbes = [1,2,3,4,5]
sqr_num = []

for n in numbes:
    sqr_num.append(n*n)

print(sqr_num)
```

```python
sqr_num = [n*n for n in numbes]
sqr_num
```

```python
import time

long_list = range(1000000)
sqr_list = []

start = time.time()
for n in long_list:
    sqr_list.append(n*n)
end = time.time()
print("Time taken by for loop:", (end-start))
```

```python
start = time.time()
sqr_list = [n*n for n in long_list]
end = time.time()
print("time taken by list comprehension:", (end-start))
```

```python
price = [100,120,80,50]
earning_per_share = [9,15,40,2]
```

```python
zip(price,earning_per_share)
```

```python
pe = [p/e for p,e in zip(price,earning_per_share)]
pe
```

```python
price = [100,120,80,50]
earning_per_share = [0,15,40,2]
pe = [p/e if e!=0 else -1 for p,e in zip(price,earning_per_share)]
pe
```

```python
prices = [120,99,150,80,200]
new_price = [price*0.9 if price > 100 else price for price in prices]
new_price
```

```python
names = ["Alice","Bob","Charlie"]
ages = [25, 30, 35]

for i in range(len(names)):
    print(names[i])
    print(ages[i])
```

```python
names = ["Alice","Bob","Charlie"]
ages = [25, 30, 35]

name_age_dict = {names[i]:ages[i] for i in range(len(names))}
name_age_dict
```
