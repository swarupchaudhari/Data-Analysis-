```python
transaction  = [
    {"customer_name": "mera","amount": 56, "data": "2024-08-01"},
    {"customer_name": "mera","amount": 12.3, "data": "2024-09-05"},
    {"customer_name": "mohan","amount": 21, "data": "2024-09-03"},
    {"customer_name": "mohan","amount": 36.8, "data": "2024-09-10"},
    {"customer_name": "lucy","amount": 66.3, "data": "2024-09-10"},
    ]
```

```python
customers = set()

for trans in transaction:
    customers.add(trans['customer_name'])

customers
```

```python
customers_list = list(customers)
customers_list
```
