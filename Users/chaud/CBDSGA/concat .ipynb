```python
import pandas as pd
```

```python
india_weather = pd.DataFrame({
    "city": ["mumbai","delhi","banglore"],
    "temperature": [32,45,30],
    "humidity": [80,60,78]
})
india_weather
```

```python
us =  pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
    "humidity": [68,65,75]
})
us
```

```python
df = pd.concat([india_weather, us])
df
```

```python
df = pd.concat([india_weather, us], keys = ["india", "us"])
df
```

```python
df.loc["india"]
```

```python
temp = pd.DataFrame({
    "city": ["dehli","mumbai","banglore"],
    "temperature": [32,45,30],
})
temp
    
```

```python
windspeed = pd.DataFrame({
    "city": ["dehli","mumbai"],
    "windspeed": [7,12],
}, index=[1,0])
windspeed
```

```python
pd.concat([temp,windspeed],axis=1)
```

```python
df2 =  pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "temperature": [65,68,71],
})
df2
```

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando","baltimore"],
    "humidity": [21,14,35, 38]
})
df1
```

```python
df3 = pd.merge(df1, df2, on="city")
df3
```

```python
df3 = pd.merge(df1, df2, on="city")
df3
```

```python
df3 = pd.merge(df1, df2, on="city", how="inner")
df3
```

```python
df3 = pd.merge(df1, df2, on="city", how="outer")
df3
```

```python
df3 = pd.merge(df1, df2, on="city", how="left")
df3
```

```python
df3 = pd.merge(df1, df2, on="city", how="right")
df3
```
